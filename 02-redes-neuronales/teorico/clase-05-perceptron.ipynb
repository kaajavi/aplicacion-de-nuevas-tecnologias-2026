## 1. El Perceptrón: la neurona artificial

```python
import numpy as np
import matplotlib.pyplot as plt

# Un perceptrón simple
# Predecir si un alumno aprueba basándose en:
#   x1 = horas de estudio
#   x2 = asistencia (%)
#   x3 = nota del parcial anterior

# Datos de un alumno
x = np.array([6, 0.85, 7])

# Pesos (determinan la importancia de cada input)
w = np.array([0.5, 0.3, 0.2])

# Bias
b = -4.0

# Suma ponderada: z = x1*w1 + x2*w2 + x3*w3 + b
z = np.dot(x, w) + b

# Activación escalón (perceptrón original)
output = 1 if z >= 0 else 0

print("=" * 50)
print("PERCEPTRÓN: ¿Aprueba el alumno?")
print("=" * 50)
print(f"Inputs:  horas={x[0]}, asistencia={x[1]}, nota_ant={x[2]}")
print(f"Pesos:   w1={w[0]}, w2={w[1]}, w3={w[2]}")
print(f"Bias:    b={b}")
print(f"\nSuma ponderada: z = {x[0]}*{w[0]} + {x[1]}*{w[1]} + {x[2]}*{w[2]} + ({b})")
print(f"                z = {x[0]*w[0]:.2f} + {x[1]*w[1]:.3f} + {x[2]*w[2]:.2f} + ({b}) = {z:.3f}")
print(f"\nActivación (escalón): z >= 0 → {z:.3f} >= 0 → {output}")
print(f"\nResultado: {'APRUEBA' if output == 1 else 'DESAPRUEBA'}")
```

```python
# Probemos con varios alumnos

alumnos = np.array([
    [6, 0.85, 7],   # buen alumno
    [2, 0.40, 4],   # mal alumno
    [4, 0.70, 5],   # alumno promedio
    [8, 0.95, 8],   # excelente alumno
    [1, 0.20, 3],   # alumno que no estudia
])

nombres = ["Juan", "María", "Pedro", "Ana", "Luis"]

print(f"{'Alumno':<10} {'Horas':>6} {'Asist':>6} {'Nota':>5} {'z':>8} {'Resultado':>10}")
print("-" * 55)
for nombre, alumno in zip(nombres, alumnos):
    z = np.dot(alumno, w) + b
    output = 1 if z >= 0 else 0
    resultado = "Aprueba" if output == 1 else "Desaprueba"
    print(f"{nombre:<10} {alumno[0]:>6.0f} {alumno[1]:>6.0%} {alumno[2]:>5.0f} {z:>8.3f} {resultado:>10}")
```

## 2. El perceptrón como clasificador lineal

```python
# Visualizar qué hace un perceptrón: traza una LÍNEA que separa dos clases

np.random.seed(42)

# Generar datos de alumnos que aprueban y desaprueban
# Aprueban: más horas y mejor nota
aprobados = np.random.randn(30, 2) * [1.5, 1] + [6, 7]
desaprobados = np.random.randn(30, 2) * [1.5, 1] + [2, 4]

plt.figure(figsize=(8, 6))
plt.scatter(aprobados[:, 0], aprobados[:, 1], color="#2ecc71", s=60, label="Aprueba", alpha=0.7)
plt.scatter(desaprobados[:, 0], desaprobados[:, 1], color="#e74c3c", s=60, label="Desaprueba", alpha=0.7)

# Línea de decisión del perceptrón: w1*x1 + w2*x2 + b = 0
# → x2 = -(w1/w2)*x1 - (b/w2)
w1, w2, bias = 0.5, 0.3, -4.0
x_line = np.linspace(0, 10, 100)
y_line = -(w1/w2) * x_line - (bias/w2)

plt.plot(x_line, y_line, color="black", linewidth=2, linestyle="--", label="Frontera de decisión")
plt.fill_between(x_line, y_line, 12, alpha=0.05, color="green")
plt.fill_between(x_line, y_line, 0, alpha=0.05, color="red")

plt.xlabel("Horas de estudio")
plt.ylabel("Nota del parcial anterior")
plt.title("El perceptrón traza una LÍNEA RECTA para separar las clases")
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.tight_layout()
plt.show()

print("El perceptrón es un CLASIFICADOR LINEAL: solo puede separar datos con una línea recta.")
```

## 3. Compuertas lógicas: AND, OR, XOR

```python
# Un perceptrón puede resolver AND y OR, pero NO puede resolver XOR

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

datos = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# AND
and_labels = np.array([0, 0, 0, 1])
for i, (punto, label) in enumerate(zip(datos, and_labels)):
    color = "#2ecc71" if label == 1 else "#e74c3c"
    axes[0].scatter(punto[0], punto[1], color=color, s=200, zorder=5,
                    edgecolor="black", linewidth=2)
    axes[0].annotate(str(label), (punto[0]+0.05, punto[1]+0.05), fontsize=14, fontweight="bold")
# Línea de separación
x_sep = np.linspace(-0.5, 1.5, 100)
axes[0].plot(x_sep, 1.5 - x_sep, "k--", linewidth=2)
axes[0].set_title("AND (separable)", fontsize=14)
axes[0].set_xlim(-0.5, 1.5)
axes[0].set_ylim(-0.5, 1.5)
axes[0].grid(True, alpha=0.3)

# OR
or_labels = np.array([0, 1, 1, 1])
for i, (punto, label) in enumerate(zip(datos, or_labels)):
    color = "#2ecc71" if label == 1 else "#e74c3c"
    axes[1].scatter(punto[0], punto[1], color=color, s=200, zorder=5,
                    edgecolor="black", linewidth=2)
    axes[1].annotate(str(label), (punto[0]+0.05, punto[1]+0.05), fontsize=14, fontweight="bold")
axes[1].plot(x_sep, 0.5 - x_sep, "k--", linewidth=2)
axes[1].set_title("OR (separable)", fontsize=14)
axes[1].set_xlim(-0.5, 1.5)
axes[1].set_ylim(-0.5, 1.5)
axes[1].grid(True, alpha=0.3)

# XOR
xor_labels = np.array([0, 1, 1, 0])
for i, (punto, label) in enumerate(zip(datos, xor_labels)):
    color = "#2ecc71" if label == 1 else "#e74c3c"
    axes[2].scatter(punto[0], punto[1], color=color, s=200, zorder=5,
                    edgecolor="black", linewidth=2)
    axes[2].annotate(str(label), (punto[0]+0.05, punto[1]+0.05), fontsize=14, fontweight="bold")
axes[2].set_title("XOR (NO separable con una línea)", fontsize=14, color="#e74c3c")
axes[2].set_xlim(-0.5, 1.5)
axes[2].set_ylim(-0.5, 1.5)
axes[2].grid(True, alpha=0.3)
axes[2].text(0.5, -0.3, "No hay ninguna línea recta\nque separe los 0s de los 1s",
             ha="center", fontsize=10, color="#e74c3c")

plt.suptitle("Limitación del Perceptrón (Minsky & Papert, 1969)", fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```

## 4. Multi-Layer Perceptron (MLP): resolviendo XOR con PyTorch

```python
import torch
import torch.nn as nn

# Datos XOR
X = torch.tensor([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=torch.float32)
y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)

# MLP: 2 entradas → 4 neuronas ocultas → 1 salida
model = nn.Sequential(
    nn.Linear(2, 4),     # capa oculta
    nn.ReLU(),           # activación no lineal
    nn.Linear(4, 1),     # capa de salida
    nn.Sigmoid()         # output entre 0 y 1
)

# Entrenamiento
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

losses = []
for epoch in range(2000):
    output = model(X)
    loss = criterion(output, y)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    losses.append(loss.item())
    if epoch % 500 == 0:
        print(f"Epoch {epoch:>4d}: Loss = {loss.item():.6f}")

# Resultado
print("\nResultados del MLP para XOR:")
with torch.no_grad():
    preds = model(X)
    for i in range(4):
        print(f"  Input: {X[i].tolist()} → Predicción: {preds[i].item():.4f} → Redondeado: {preds[i].round().item():.0f} (esperado: {y[i].item():.0f})")
```

```python
# Visualizar el aprendizaje y la frontera de decisión aprendida

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Curva de loss
axes[0].plot(losses, color="steelblue", linewidth=1.5)
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].set_title("Curva de entrenamiento")
axes[0].grid(True, alpha=0.3)

# Frontera de decisión aprendida
xx, yy = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))
grid = torch.tensor(np.c_[xx.ravel(), yy.ravel()], dtype=torch.float32)
with torch.no_grad():
    zz = model(grid).numpy().reshape(xx.shape)

axes[1].contourf(xx, yy, zz, levels=50, cmap="RdYlGn", alpha=0.7)
axes[1].contour(xx, yy, zz, levels=[0.5], colors="black", linewidths=2)

for i in range(4):
    color = "#2ecc71" if y[i].item() == 1 else "#e74c3c"
    axes[1].scatter(X[i, 0], X[i, 1], color=color, s=200, zorder=5,
                    edgecolor="black", linewidth=2)

axes[1].set_title("Frontera de decisión del MLP para XOR")
axes[1].set_xlabel("x1")
axes[1].set_ylabel("x2")

plt.tight_layout()
plt.show()

print("La frontera de decisión ya NO es una línea recta.")
print("Las capas ocultas permiten aprender fronteras curvas y complejas.")
```

## 5. Funciones de Activación

```python
import torch.nn.functional as F

z = torch.linspace(-5, 5, 200)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# ReLU
axes[0, 0].plot(z.numpy(), F.relu(z).numpy(), color="steelblue", linewidth=2)
axes[0, 0].set_title("ReLU: max(0, z)", fontsize=13)
axes[0, 0].axhline(y=0, color="gray", linestyle="--", alpha=0.3)
axes[0, 0].axvline(x=0, color="gray", linestyle="--", alpha=0.3)
axes[0, 0].set_xlabel("z")
axes[0, 0].set_ylabel("ReLU(z)")
axes[0, 0].text(1, 4, "Default para\ncapas ocultas", fontsize=11, color="steelblue")

# Sigmoid
axes[0, 1].plot(z.numpy(), torch.sigmoid(z).numpy(), color="#e74c3c", linewidth=2)
axes[0, 1].axhline(y=0.5, color="gray", linestyle="--", alpha=0.3)
axes[0, 1].set_title("Sigmoid: 1 / (1 + e^(-z))", fontsize=13)
axes[0, 1].set_xlabel("z")
axes[0, 1].set_ylabel("σ(z)")
axes[0, 1].text(1, 0.3, "Rango (0, 1)\nPara clasificación\nbinaria", fontsize=11, color="#e74c3c")

# Tanh
axes[1, 0].plot(z.numpy(), torch.tanh(z).numpy(), color="#2ecc71", linewidth=2)
axes[1, 0].axhline(y=0, color="gray", linestyle="--", alpha=0.3)
axes[1, 0].set_title("Tanh: (e^z - e^(-z)) / (e^z + e^(-z))", fontsize=13)
axes[1, 0].set_xlabel("z")
axes[1, 0].set_ylabel("tanh(z)")
axes[1, 0].text(1, -0.5, "Rango (-1, 1)\nCentrado en 0", fontsize=11, color="#2ecc71")

# Softmax (ejemplo con 3 clases)
logits = torch.linspace(-3, 3, 200)
# Simular 3 clases variando un logit
softmax_vals = []
for val in logits:
    s = F.softmax(torch.tensor([val.item(), 1.0, 0.5]), dim=0)
    softmax_vals.append(s.numpy())
softmax_vals = np.array(softmax_vals)

axes[1, 1].plot(logits.numpy(), softmax_vals[:, 0], linewidth=2, label="Clase 0")
axes[1, 1].plot(logits.numpy(), softmax_vals[:, 1], linewidth=2, label="Clase 1")
axes[1, 1].plot(logits.numpy(), softmax_vals[:, 2], linewidth=2, label="Clase 2")
axes[1, 1].set_title("Softmax: probabilidades que suman 1", fontsize=13)
axes[1, 1].set_xlabel("Logit de clase 0")
axes[1, 1].set_ylabel("Probabilidad")
axes[1, 1].legend()
axes[1, 1].text(0.5, 0.8, "Para clasificación\nmulticlase", fontsize=11, color="gray")

for ax in axes.flat:
    ax.grid(True, alpha=0.3)

plt.suptitle("Funciones de Activación", fontsize=15, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# Demo numérica de cada activación

z = torch.tensor([-2.0, -1.0, 0.0, 1.0, 2.0])

print(f"z:        {z.tolist()}")
print(f"ReLU:     {F.relu(z).tolist()}")
print(f"Sigmoid:  {[round(x, 3) for x in torch.sigmoid(z).tolist()]}")
print(f"Tanh:     {[round(x, 3) for x in torch.tanh(z).tolist()]}")

print(f"\nSoftmax ejemplo:")
logits = torch.tensor([2.0, 1.0, 0.1])
probs = F.softmax(logits, dim=0)
print(f"  Logits: {logits.tolist()}")
print(f"  Softmax: {[round(x, 3) for x in probs.tolist()]}")
print(f"  Suma:    {probs.sum().item():.3f} (siempre suman 1)")
print(f"  → Clase con mayor probabilidad: {probs.argmax().item()} ({probs.max().item():.1%})")
```

## 6. Forward Pass: cómo fluyen los datos

```python
# Forward pass manual con NumPy
# Red: 3 inputs → 2 neuronas ocultas (ReLU) → 1 salida (Sigmoid)

print("=" * 60)
print("FORWARD PASS PASO A PASO")
print("=" * 60)

# Input: un alumno [horas_estudio=6, asistencia=0.85, nota=7]
X = np.array([[6, 0.85, 7]])  # shape (1, 3)
print(f"\nInput X: {X} — shape {X.shape}")

# Capa oculta: 3 inputs → 2 neuronas
W1 = np.array([[ 0.3, -0.1],
                [ 0.5,  0.8],
                [-0.2,  0.4]])  # shape (3, 2)
b1 = np.array([0.1, -0.3])     # shape (2,)

print(f"\n--- Capa Oculta ---")
print(f"W1: shape {W1.shape}")
print(f"b1: {b1}")

# Paso 1: Suma ponderada
Z1 = X @ W1 + b1  # (1,3) @ (3,2) + (2,) = (1,2)
print(f"Z1 = X @ W1 + b1 = {Z1}")

# Paso 2: Activación ReLU
A1 = np.maximum(0, Z1)
print(f"A1 = ReLU(Z1) = {A1}")

# Capa de salida: 2 → 1
W2 = np.array([[0.6],
                [0.9]])  # shape (2, 1)
b2 = np.array([-0.5])   # shape (1,)

print(f"\n--- Capa de Salida ---")
print(f"W2: shape {W2.shape}")
print(f"b2: {b2}")

# Paso 3: Suma ponderada
Z2 = A1 @ W2 + b2  # (1,2) @ (2,1) + (1,) = (1,1)
print(f"Z2 = A1 @ W2 + b2 = {Z2}")

# Paso 4: Activación Sigmoid
A2 = 1 / (1 + np.exp(-Z2))
print(f"A2 = Sigmoid(Z2) = {A2}")

print(f"\nPredicción: {A2[0][0]:.4f} → {'Aprueba' if A2[0][0] > 0.5 else 'Desaprueba'}")
```

```python
# Lo mismo en PyTorch (mucho más simple)

class MiRed(nn.Module):
    def __init__(self):
        super().__init__()
        self.capa1 = nn.Linear(3, 2)   # 3 inputs → 2 ocultas
        self.capa2 = nn.Linear(2, 1)   # 2 ocultas → 1 salida

    def forward(self, x):
        x = torch.relu(self.capa1(x))     # capa oculta + ReLU
        x = torch.sigmoid(self.capa2(x))  # salida + Sigmoid
        return x

modelo = MiRed()
entrada = torch.tensor([[6.0, 0.85, 7.0]])
prediccion = modelo(entrada)  # esto ejecuta forward()

print(f"PyTorch - Predicción: {prediccion.item():.4f}")
print(f"\nParámetros del modelo:")
for name, param in modelo.named_parameters():
    print(f"  {name}: shape {list(param.shape)}, valores: {param.data.numpy().flatten()[:4]}...")

total_params = sum(p.numel() for p in modelo.parameters())
print(f"\nTotal de parámetros: {total_params}")
```

## 7. Ejemplo práctico: MLP para clasificación de dígitos (MNIST)

```python
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader

# Cargar MNIST (dígitos escritos a mano, 28x28 píxeles)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

train_data = torchvision.datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_data = torchvision.datasets.MNIST(root='./data', train=False, download=True, transform=transform)

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64)

print(f"Dataset MNIST:")
print(f"  Training: {len(train_data)} imágenes")
print(f"  Test:     {len(test_data)} imágenes")
print(f"  Tamaño:   28x28 píxeles, escala de grises")
print(f"  Clases:   dígitos 0-9")

# Visualizar algunas muestras
fig, axes = plt.subplots(2, 8, figsize=(14, 4))
for i, ax in enumerate(axes.flat):
    img, label = train_data[i]
    ax.imshow(img.squeeze(), cmap="gray")
    ax.set_title(f"Label: {label}")
    ax.axis("off")
plt.suptitle("Muestras del dataset MNIST", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# MLP para clasificar dígitos
# 28*28 = 784 inputs → 128 ocultas → 64 ocultas → 10 salidas (dígitos 0-9)

mlp = nn.Sequential(
    nn.Flatten(),           # 28x28 → 784
    nn.Linear(784, 128),
    nn.ReLU(),
    nn.Linear(128, 64),
    nn.ReLU(),
    nn.Linear(64, 10)       # 10 clases (softmax implícito en CrossEntropyLoss)
)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(mlp.parameters(), lr=0.001)

# Entrenamiento
num_epochs = 5
for epoch in range(num_epochs):
    mlp.train()
    total_loss = 0
    correct = 0
    total = 0
    
    for images, labels in train_loader:
        outputs = mlp(images)
        loss = criterion(outputs, labels)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
    
    acc = 100. * correct / total
    print(f"Epoch {epoch+1}/{num_epochs} - Loss: {total_loss/len(train_loader):.4f} - Accuracy: {acc:.1f}%")

# Evaluar en test
mlp.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        outputs = mlp(images)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

print(f"\nAccuracy en test: {100.*correct/total:.1f}%")
print(f"Total de parámetros: {sum(p.numel() for p in mlp.parameters()):,}")
```

```python
# Ver predicciones individuales

fig, axes = plt.subplots(2, 8, figsize=(14, 4))
mlp.eval()
with torch.no_grad():
    for i, ax in enumerate(axes.flat):
        img, label = test_data[i]
        pred = mlp(img.unsqueeze(0)).argmax(1).item()
        color = "green" if pred == label else "red"
        ax.imshow(img.squeeze(), cmap="gray")
        ax.set_title(f"Pred: {pred}", color=color, fontweight="bold")
        ax.axis("off")

plt.suptitle("Predicciones del MLP (verde=correcto, rojo=incorrecto)", fontsize=14)
plt.tight_layout()
plt.show()
```

## Resumen

| Concepto | Clave |
|----------|-------|
| Perceptrón | `output = activación(x·w + b)` |
| Clasificador lineal | Solo separa con una línea recta |
| XOR | Un perceptrón NO puede resolverlo |
| MLP | Capas ocultas → fronteras no lineales |
| ReLU | Default para capas ocultas |
| Sigmoid | Salida binaria (0-1) |
| Softmax | Salida multiclase (probabilidades) |
| Forward pass | Datos × Pesos + Bias → Activación |
